```python
def srednia(lista):
  sum=0
  for i in lista:
    sum+=i
  sum=sum/len(lista)
  print(sum)
```

```python
pajaki = [48, 49, 51, 52]
ender_dragony = [0, 0, 100, 100]
szkielety = [10, 30, 70, 90]
zombie = [49, 50, 50, 51]

import statistics
import matplotlib.pyplot as plt

def warjancja(lista):
  srednia = statistics.mean(lista)
  war = sum((x-srednia) ** 2 for x in lista)/len(lista)
  print(war)

warjancja(pajaki)
warjancja(ender_dragony)
warjancja(szkielety)
warjancja(zombie)
```
